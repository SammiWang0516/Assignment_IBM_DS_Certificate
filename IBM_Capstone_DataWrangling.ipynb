<h3 style = 'color: yellow'>task3: calculate the number and occurence of mission outcome of the orbits</h3>

```python
landing_outcomes = df['Outcome'].value_counts()
landing_outcomes
```

```python
for i, outcome in enumerate(landing_outcomes.keys()):
    print(i, outcome)
```

```python
bad_outcomes = set(landing_outcomes.keys()[[1, 3, 5, 6, 7]])
bad_outcomes
```

<h3 style = 'color: yellow'>task4: create a landing outcome label from outcome column</h3>

```python
landing_class = []
for index, row in df.iterrows():
    if row['Outcome'] in bad_outcomes:
        landing_class.append(0)
    else:
        landing_class.append(1)
len(landing_class)
```

```python
df['Class'] = landing_class
df[['Class']].head(8)
```
